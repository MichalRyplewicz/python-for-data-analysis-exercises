```python
sequence = [1, 2, None, 4, None, 5]
total = 0
for xd in sequence:
    if xd is None:
        continue
    total += xd

print(total)
```

```python
sequence2 = [1, 2, 0, 4, 6, 5, 2, 1]
total2 = 0
for val in sequence2:
    if val == 5:
        break
    total2 += val

print(total2)
```

```python
x = 256
total = 0
while x > 0:
    if total > 500:
        break
    total += x
    x = x // 2
    print(total)

```
