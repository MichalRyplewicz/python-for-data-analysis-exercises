```python
words = ["apple", "bat", "bar", "atom", "book"]
by_letter = {}

for word in words:
    letter = word[0]
    if letter not in by_letter:
        by_letter[letter] = [word]
    else:
        by_letter[letter].append(word)

by_letter
```

```python
# set default daj mi listę pod tą literą, a jeśli jej nie ma, najpierw załóż pustą [] - wartość defaultowa
#.append dołóż słowo do tej listy
words = ["like", "mike", "mind", "side", "snake"]
by_letter = {}
for word in words:
    letter = word[0]
    by_letter.setdefault(letter,[]).append(word)

by_letter
```

```python

from collections import defaultdict

by_letter = defaultdict(list) # warunek na poczatku, ze jeżeli pojawi sie klucz którego nie ma, załóż go z pusta listą
for word in words:
    by_letter[word[0]].append(word)

by_letter
```

```python
produkty = ["chleb", "mleko", "maslo"]
ceny = [4.50, 3.20, 9.90]

cennik = dict(zip(produkty, ceny))

for produkt, cena in cennik.items():
    print(f"{produkt}: {cena:.2f} zł")
```

```python
ceny_check = ["mleko", "ser"]
for produkt in ceny_check:
    cena1 = cennik.get(produkt,0)
    print(f"{produkt}: {cena1:.2f} zł")
```

```python
miasta = ["Gdańsk", "Gdynia", "Sopot", "Słupsk", "Kraków", "Gniezno", "Katowice"]
slownik = {}

for miasto in miasta:
    letter = miasto[0]
    slownik.setdefault(letter, []).append(miasto)

slownik
```

```python
miasta = ["Gdańsk", "Gdynia", "Sopot", "Słupsk", "Kraków", "Gniezno", "Katowice"]
slownik_po_literze = {}

for miasto in miasta:
    litera_1 = miasto[0]
    slownik_po_literze.setdefault(litera_1, []).append(miasto)

slownik_po_literze  
```
